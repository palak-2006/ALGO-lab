```python
# Exercise 1
import time
def func1(a, b):
  result = 0
  for i in range(0, b):
    result += a
  return result
```

```python
# Sample function call: func1(5, 3)
print(func1(5,3))
#the function performs multiplication using repeated addition
```

```python

start = time.perf_counter()
func1(5,3)
end = time.perf_counter()
print("elapsed time = ",(end-start)*1000)
```

```python
# Exercise 2
def func2(a, b):
  result = 1
  while b > 0:
    result = result * a
    b = b - 1
  return result
```

```python
# Sample function call: func2(2, 4)
x = print(func2(2,4))
#the function performs power like a to the power b using repeated multiplication
```

```python
import time
start = time.perf_counter()
func2(2,4)
end = time.perf_counter()
print("elapsed time = ",(end-start)*1000)
```

```python
# Exercise 4
def func4(a, b):
    b.sort() # Time complexity for this function is O(Nlog N) where N = len(b)
    intersect = 0
    for x in a:
        if binary_search(b, x) >= 0:
            intersect = intersect + 1
    return intersect

def binary_search(arr, target):
    low, high = 0, len(arr) - 1
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1
```

```python
# Sample function call: func4([1, 2, 3, 4], [2, 4, 6, 8])
func4([1, 2, 3, 4], [2, 4, 6, 8])
```

```python
# Exercise 5
def func5(n):
    guess = 1
    while guess * guess <= n:
        if guess * guess == n:
            return guess
        guess = guess + 1
    return None
```

```python
# Sample function call: func5(16)
print(func5(16))
#this function perform the perfect square root of a no. if there is no perfect square root of that number it show none
```

```python
import time
start = time.perf_counter()
func5(16)
end = time.perf_counter()
print("elapsed time = ",(end-start)*1000)
```

```python
# Exercise 6
def func8(A):
    func8_helper(A, 0, len(A) -1)

def func8_helper(A, i, j):
    #Input: An array A and two indices i and j.
    n = j - 1 + 1
    if n ==2:
        if A[i] > A[j]:
            # Swap A[i] and A[j]
            A[i], A[j] = A[j], A[i]
        elif n > 2:
            m = n // 3
            func8_helper(A, i, j - m)
            func8_helper(A, i + m, j)
            func8_helper(A, i, j - m)
        return A
```

```python
#Sample function call: func8([3,1,2,7,8,6])
print(func8([3,1,2,7,8,6]))
```

```python
# Exercise 7
def func12(arr, k):
    arr_len = len(arr)
    k = k % arr_len
    # Initialize a new array of the same length as 'arr'
    custom_arr = [0] * arr_len

    for i in range(arr_len):
        mystery_index = (i + k) % arr_len
        custom_arr[mystery_index] = arr[i]
    return custom_arr

```

```python
#Sample function call: func12([1, 2, 3, 4, 5], 2)
print(func12([1, 2, 3, 4, 5], 2))
# this function sort the last k elements of array
```

```python
import time
start = time.perf_counter()
func12([1, 2, 3, 4, 5], 2)
end = time.perf_counter()
print("elapsed time = ",(end-start)*1000)
```

Programs on 1-d arrays like - sum of elements of array, searching an element inarray, finding minimum and maximum element in array, count the number ofeven and odd numbers in array. For all such programs, also find the timeComplexity, compare if there are multiple methods

```python
# write a function named as sum_arr to find sum of elements in given array
def sum_arr(arr):
  n = len(arr)
  sum = 0
  for i in range(n):
    sum = sum + arr[i]
  return sum
```

```python
#call the function
#arr = [12, 3, 4, 15,45,78,450]
#print(sum_arr)
arr = [12, 3, 4, 15,45,78,450]
print(sum_arr(arr))
```

```python
import time
start = time.perf_counter()
sum_arr(arr)
end = time.perf_counter()
print("elapsed time = ",(end-start)*1000)
```

```python
#write a function named as max_min to find minimum (or maximum) element in an array
def max_min(arr):
  max = arr[0]
  min = arr[0]
  n = len(arr)
  for i in range(n):
    if(arr[i]>max):
      max = arr[i]
    if(arr[i]<min):
      min = arr[i]
  return max,min
print(max_min([1,2,3]))
```

```python
# write a function named as Counting_EvenOdd to count number of even and odd elements in an array
def Counting_EvenOdd(arr):
  c1 = 0
  c2 = 0
  n = len(arr)
  for i in range(n):
    if(arr[i]%2 == 0):
      c1 = c1+1
    else:
      c2 = c2+1
  return c1 , c2
```
